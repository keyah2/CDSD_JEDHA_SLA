L'objectif est de construire un modèle capable de classifier chaque SMS comme **spam** ou **ham** (message normal), en se basant **uniquement sur le contenu textuel** du message.

## 1. Imports et configuration

```python
# --- Bibliothèques générales ---
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

# --- Scikit-learn : split et métriques ---
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, classification_report, ConfusionMatrixDisplay
)

# --- TensorFlow / Keras : deep learning ---
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.callbacks import EarlyStopping

# --- TensorFlow Hub : transfer learning ---
import tensorflow_hub as hub

# Fixer la graine aléatoire pour la reproductibilité
SEED = 42
tf.random.set_seed(SEED)
np.random.seed(SEED)

print(f"TensorFlow version : {tf.__version__}")
print(f"TensorFlow Hub version : {hub.__version__}")
```

---
## 2. Chargement et exploration des données

```python
# Chargement du fichier CSV (encodage latin1 requis pour ce dataset)
df_raw = pd.read_csv('../data/spam.csv', encoding='latin1')

# Affichage des premières lignes pour comprendre la structure brute
print("Aperçu brut du dataset :")
df_raw.head()
```

```python
# Le dataset contient des colonnes inutiles (Unnamed: 2, 3, 4)
# On ne garde que les deux colonnes utiles et on les renomme
df = df_raw[['v1', 'v2']].rename(columns={'v1': 'label', 'v2': 'text'})

# Encodage des labels : ham -> 0, spam -> 1
df['label'] = df['label'].map({'ham': 0, 'spam': 1})

print(f"Taille du dataset : {df.shape[0]} messages")
print(f"\nRépartition des classes :")
print(df['label'].value_counts().rename({0: 'ham', 1: 'spam'}))
df.head(10)
```

```python
# Vérification des valeurs manquantes
print("Valeurs manquantes par colonne :")
print(df.isnull().sum())

# Longueur des messages
df['longueur'] = df['text'].apply(len)
print(f"\nLongueur moyenne des messages : {df['longueur'].mean():.0f} caractères")
print(f"Longueur max : {df['longueur'].max()} | Longueur min : {df['longueur'].min()}")
```

```python
# --- Visualisation 1 : Répartition des classes ---
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Diagramme en barres
counts = df['label'].value_counts()
axes[0].bar(['ham (0)', 'spam (1)'], counts.values, color=['steelblue', 'tomato'])
axes[0].set_title('Répartition des classes')
axes[0].set_ylabel('Nombre de messages')
for i, v in enumerate(counts.values):
    axes[0].text(i, v + 30, str(v), ha='center', fontweight='bold')

# Distribution des longueurs par classe
axes[1].hist(df[df['label'] == 0]['longueur'], bins=50, alpha=0.6, label='ham', color='steelblue')
axes[1].hist(df[df['label'] == 1]['longueur'], bins=50, alpha=0.6, label='spam', color='tomato')
axes[1].set_title('Distribution de la longueur des messages')
axes[1].set_xlabel('Nombre de caractères')
axes[1].set_ylabel('Fréquence')
axes[1].legend()

plt.tight_layout()
plt.show()

print("\nObservation : le dataset est déséquilibré (~87% ham / ~13% spam).")
print("Les messages spam sont en général plus longs que les messages ham.")
```

```python
# Exemples de messages ham et spam pour mieux comprendre les données
print("=== Exemples de messages HAM ===")
for msg in df[df['label'] == 0]['text'].sample(3, random_state=SEED).values:
    print(f"  - {msg}")

print("\n=== Exemples de messages SPAM ===")
for msg in df[df['label'] == 1]['text'].sample(3, random_state=SEED).values:
    print(f"  - {msg}")
```

## 3. Prétraitement des données

Le deep learning sur texte requiert de transformer les chaînes de caractères en séquences numériques.  
Voici les étapes :
1. Séparation en jeu d'entraînement et jeu de test
2. Tokenization (construction du vocabulaire)
3. Conversion des textes en séquences d'entiers
4. Padding (uniformisation de la longueur)

```python
# Séparation des variables explicatives (X) et de la cible (y)
X = df['text'].values
y = df['label'].values

# Division train / test avec stratification pour conserver les proportions de classes
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=SEED,
    stratify=y  # garantit la même proportion spam/ham dans train et test
)

print(f"Taille du jeu d'entraînement : {len(X_train)} messages")
print(f"Taille du jeu de test        : {len(X_test)} messages")
print(f"\nProportion spam dans train   : {y_train.mean():.2%}")
print(f"Proportion spam dans test    : {y_test.mean():.2%}")
```

```python
# Paramètres du prétraitement
VOCAB_SIZE = 5000     # taille maximale du vocabulaire
MAX_LEN    = 50       # longueur maximale de chaque séquence (en tokens)
OOV_TOKEN  = '<OOV>'  # token pour les mots hors-vocabulaire

# Tokenization : on apprend le vocabulaire sur le jeu d'entraînement UNIQUEMENT
# (jamais sur le test, pour éviter la fuite de données)
tokenizer = Tokenizer(num_words=VOCAB_SIZE, oov_token=OOV_TOKEN)
tokenizer.fit_on_texts(X_train)

print(f"Taille réelle du vocabulaire appris : {len(tokenizer.word_index)} mots")
print(f"Vocabulaire limité à                : {VOCAB_SIZE} mots")
```

```python
# Conversion des textes en séquences d'entiers
X_train_seq = tokenizer.texts_to_sequences(X_train)
X_test_seq  = tokenizer.texts_to_sequences(X_test)

# Exemple : visualisation d'un message avant/après tokenization
print("Message original :")
print(f"  {X_train[0]}")
print("\nAprès tokenization :")
print(f"  {X_train_seq[0]}")
```

```python
# Padding : uniformisation de la longueur de toutes les séquences à MAX_LEN
# padding='post'    -> les zéros sont ajoutés à la fin
# truncating='post' -> les séquences trop longues sont coupées à la fin
X_train_pad = pad_sequences(X_train_seq, maxlen=MAX_LEN, padding='post', truncating='post')
X_test_pad  = pad_sequences(X_test_seq,  maxlen=MAX_LEN, padding='post', truncating='post')

print(f"Forme de X_train après padding : {X_train_pad.shape}")
print(f"Forme de X_test après padding  : {X_test_pad.shape}")
print("\nExemple de séquence paddée :")
print(f"  {X_train_pad[0]}")
```

## 4. Modèle 1 — Embedding from scratch + GlobalAveragePooling1D

Ce premier modèle est le **baseline** : simple, rapide à entraîner, et souvent très efficace sur des textes courts.  
L'embedding est appris directement depuis les données (pas de pré-entraînement).  
Le `GlobalAveragePooling1D` agrège les représentations de chaque mot en une représentation globale du message.

```python
# Paramètres communs aux deux premiers modèles
EMBEDDING_DIM = 32   # dimension de l'espace d'embedding
BATCH_SIZE    = 32
EPOCHS        = 20

# Callback EarlyStopping : arrêt anticipé si la val_loss ne s'améliore plus pendant 3 époques
early_stop = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

# --- Architecture du modèle 1 ---
model1 = models.Sequential([
    # Embedding : transforme chaque entier (token) en vecteur de dimension EMBEDDING_DIM
    layers.Embedding(input_dim=VOCAB_SIZE, output_dim=EMBEDDING_DIM, input_length=MAX_LEN),

    # GlobalAveragePooling1D : fait la moyenne des vecteurs sur toute la séquence
    layers.GlobalAveragePooling1D(),

    # Couche dense avec activation ReLU
    layers.Dense(16, activation='relu'),

    # Dropout : régularisation pour éviter le surapprentissage
    layers.Dropout(0.3),

    # Couche de sortie : sigmoid -> probabilité d'être spam (entre 0 et 1)
    layers.Dense(1, activation='sigmoid')
], name='Modele1_GAP')

model1.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model1.summary()
```

```python
# Entraînement du modèle 1
history1 = model1.fit(
    X_train_pad, y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.2,  # 20% du train utilisés pour la validation
    callbacks=[early_stop],
    verbose=1
)
```

```python
# Fonction utilitaire : afficher les courbes d'apprentissage (loss et accuracy)
def plot_history(history, titre):
    fig, axes = plt.subplots(1, 2, figsize=(13, 4))

    # Courbe de perte
    axes[0].plot(history.history['loss'],     label='Train')
    axes[0].plot(history.history['val_loss'], label='Validation')
    axes[0].set_title(f'{titre} — Perte (Loss)')
    axes[0].set_xlabel('Époque')
    axes[0].set_ylabel('Loss')
    axes[0].legend()

    # Courbe d'accuracy
    axes[1].plot(history.history['accuracy'],     label='Train')
    axes[1].plot(history.history['val_accuracy'], label='Validation')
    axes[1].set_title(f'{titre} — Accuracy')
    axes[1].set_xlabel('Époque')
    axes[1].set_ylabel('Accuracy')
    axes[1].legend()

    plt.tight_layout()
    plt.show()

plot_history(history1, 'Modèle 1 — Embedding + GlobalAveragePooling1D')
```

```python
# Fonction utilitaire : évaluer un modèle et retourner les métriques
def evaluer_modele(model, X_test, y_test, nom_modele, couleur='Blues'):
    # Prédictions
    y_proba = model.predict(X_test, verbose=0).ravel()
    y_pred  = (y_proba >= 0.5).astype(int)

    # Calcul des métriques
    acc  = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred)
    rec  = recall_score(y_test, y_pred)
    f1   = f1_score(y_test, y_pred)

    print(f"\n{'='*55}")
    print(f"  {nom_modele}")
    print(f"{'='*55}")
    print(f"  Accuracy  : {acc:.4f}")
    print(f"  Precision : {prec:.4f}")
    print(f"  Recall    : {rec:.4f}")
    print(f"  F1-score  : {f1:.4f}")
    print(f"\nRapport de classification complet :")
    print(classification_report(y_test, y_pred, target_names=['ham (0)', 'spam (1)']))

    # Matrice de confusion
    ConfusionMatrixDisplay.from_predictions(
        y_test, y_pred,
        display_labels=['ham', 'spam'],
        cmap=couleur
    )
    plt.title(f'Matrice de confusion — {nom_modele}')
    plt.show()

    return {'Modèle': nom_modele, 'Accuracy': acc, 'Precision': prec, 'Recall': rec, 'F1-score': f1}

# Évaluation du modèle 1
res1 = evaluer_modele(model1, X_test_pad, y_test, 'Modèle 1 — Embedding + GAP', couleur='Blues')
```

## 5. Modèle 2 — Embedding from scratch + LSTM

Le réseau LSTM (**Long Short-Term Memory**) est une architecture récurrente qui prend en compte **l'ordre des mots** dans la séquence, contrairement au GlobalAveragePooling1D qui fait simplement la moyenne.  
Il est théoriquement plus expressif, mais plus long à entraîner et pas toujours supérieur sur des textes courts.

```python
# Réinitialisation du callback pour ce nouveau modèle
early_stop2 = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

# --- Architecture du modèle 2 ---
model2 = models.Sequential([
    # Embedding : même principe que le modèle 1
    layers.Embedding(input_dim=VOCAB_SIZE, output_dim=EMBEDDING_DIM, input_length=MAX_LEN),

    # LSTM : modélise les dépendances séquentielles entre les tokens
    layers.LSTM(32),

    # Dropout pour régularisation
    layers.Dropout(0.3),

    # Couche dense intermédiaire
    layers.Dense(16, activation='relu'),

    # Sortie binaire
    layers.Dense(1, activation='sigmoid')
], name='Modele2_LSTM')

model2.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model2.summary()
```

```python
# Entraînement du modèle 2
history2 = model2.fit(
    X_train_pad, y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    callbacks=[early_stop2],
    verbose=1
)

plot_history(history2, 'Modèle 2 — Embedding + LSTM')
```

```python
# Évaluation du modèle 2
res2 = evaluer_modele(model2, X_test_pad, y_test, 'Modèle 2 — Embedding + LSTM', couleur='Greens')
```

## 6. Modèle 3 — Transfer Learning avec TensorFlow Hub

### Qu'est-ce que le transfer learning ?

Le **transfer learning** (apprentissage par transfert) consiste à réutiliser un modèle pré-entraîné sur de grandes quantités de données pour accélérer et améliorer l'apprentissage sur une tâche spécifique.

Ici, on utilise le modèle **NNLM-en-dim50** de TensorFlow Hub : un encodeur de texte pré-entraîné sur le corpus **Google News** (des milliards de mots). Il transforme directement un texte brut en un vecteur de 50 dimensions, sans avoir besoin de tokenizer manuellement.

**Avantages :**
- L'encodeur connaît déjà le sens des mots grâce à son entraînement massif
- Moins besoin de données d'entraînement
- Généralement plus robuste aux nouveaux textes

```python
# URL du modèle NNLM pré-entraîné sur Google News (~50MB, téléchargé automatiquement)
TFHUB_URL = "https://tfhub.dev/google/nnlm-en-dim50/2"

print("Chargement du modèle pré-entraîné depuis TensorFlow Hub...")
print(f"Modèle : {TFHUB_URL}")
print("Ce modèle a été entraîné sur des milliards de mots du corpus Google News.")
```

```python
# Réinitialisation du callback
early_stop3 = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

# --- Architecture du modèle 3 (Transfer Learning) ---
# TensorFlow Hub est conçu pour tf_keras (Keras 2), pas Keras 3
# On importe tf_keras qui est livré avec TF 2.16+ pour la compatibilité
import tf_keras

# Couche d'entrée tf_keras : compatible avec hub.KerasLayer
input_layer = tf_keras.Input(shape=[], dtype=tf.string)

# Encodeur NNLM pré-entraîné sur Google News (milliards de mots)
# trainable=True : fine-tuning, les poids pré-entraînés s'adaptent à nos données
x = hub.KerasLayer(TFHUB_URL, trainable=True, name='nnlm_encoder')(input_layer)

# Dropout pour régularisation
x = tf_keras.layers.Dropout(0.3)(x)

# Couche dense intermédiaire
x = tf_keras.layers.Dense(16, activation='relu')(x)

# Dropout supplémentaire
x = tf_keras.layers.Dropout(0.2)(x)

# Couche de sortie sigmoid
output = tf_keras.layers.Dense(1, activation='sigmoid')(x)

# Modèle fonctionnel tf_keras (compatible hub.KerasLayer)
model3 = tf_keras.Model(inputs=input_layer, outputs=output, name='Modele3_TransferLearning')

model3.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model3.summary()
```

```python
# Entraînement du modèle 3
# On utilise le EarlyStopping de tf_keras (compatible avec le modèle 3)
early_stop3_tfk = tf_keras.callbacks.EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

history3 = model3.fit(
    np.array(X_train), np.array(y_train),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    validation_split=0.2,
    callbacks=[early_stop3_tfk],
    verbose=1
)

plot_history(history3, 'Modèle 3 — Transfer Learning (NNLM)')
```

```python
# Évaluation du modèle 3 (sur les textes bruts, pas paddés)
res3 = evaluer_modele(model3, np.array(X_test), y_test, 'Modèle 3 — Transfer Learning (NNLM)', couleur='Oranges')
```

## 7. Comparaison des trois modèles

```python
# Rassemblement des résultats dans un DataFrame
resultats = pd.DataFrame([res1, res2, res3])
resultats = resultats.set_index('Modèle')

# Affichage du tableau de comparaison
print("\nTableau comparatif des performances sur le jeu de test :")
print(resultats.round(4).to_string())
resultats.round(4)
```

```python
# Visualisation comparative des métriques
metriques = ['Accuracy', 'Precision', 'Recall', 'F1-score']
x = np.arange(len(metriques))
largeur = 0.25
couleurs = ['steelblue', 'seagreen', 'darkorange']

fig, ax = plt.subplots(figsize=(12, 5))

for i, (idx, row) in enumerate(resultats.iterrows()):
    valeurs = [row[m] for m in metriques]
    barres = ax.bar(x + i * largeur, valeurs, largeur, label=idx, color=couleurs[i], alpha=0.85)
    for b in barres:
        ax.text(
            b.get_x() + b.get_width() / 2,
            b.get_height() + 0.003,
            f'{b.get_height():.3f}',
            ha='center', va='bottom', fontsize=8
        )

ax.set_title('Comparaison des performances des trois modèles', fontsize=14)
ax.set_xticks(x + largeur)
ax.set_xticklabels(metriques)
ax.set_ylabel('Score')
ax.set_ylim(0.8, 1.02)
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

```python
# Identification du meilleur modèle selon le F1-score (métrique la plus pertinente sur dataset déséquilibré)
meilleur_idx = resultats['F1-score'].idxmax()
meilleur_f1  = resultats['F1-score'].max()

print(f"\n>>> Meilleur modèle selon le F1-score : {meilleur_idx}")
print(f"    F1-score : {meilleur_f1:.4f}")
print(f"\nRappel : on privilégie le F1-score car le dataset est déséquilibré.")
print(f"L'accuracy seule est trompeuse : un modèle qui prédit toujours 'ham' aurait déjà ~87% d'accuracy.")
```

## 8. Conclusion

### Résultats obtenus

Trois modèles de deep learning ont été entraînés et évalués sur 5 572 SMS (4 457 train / 1 115 test) :

| Modèle | Accuracy | Precision | Recall | F1-score |
|---|---:|---:|---:|---:|
| **Embedding + GlobalAveragePooling1D** | **0.9857** | **0.9716** | **0.9195** | **0.9448** |
| Embedding + LSTM | 0.9659 | 0.9237 | 0.8121 | 0.8643 |
| Transfer Learning (NNLM) | 0.9587 | 0.8601 | 0.8255 | 0.8425 |

### Analyse des résultats

Le **Modèle 1 (Embedding + GlobalAveragePooling1D)** est le grand gagnant avec un **F1-score de 0.9448**, supérieur aux deux autres modèles sur toutes les métriques.

- **Modèle 1** : meilleure precision (0.9716) et meilleur recall (0.9195) — très peu de faux positifs ET de faux négatifs.
- **Modèle 2 (LSTM)** : plus complexe, mais moins performant. Le LSTM apporte peu de valeur sur des SMS courts où l'ordre des mots est moins déterminant.
- **Modèle 3 (Transfer Learning NNLM)** : contre-intuitif, mais explicable — le modèle NNLM a été pré-entraîné sur Google News, un corpus très différent du langage SMS (abréviations, fautes, codes). Le domaine source et le domaine cible sont trop éloignés pour que le transfer learning soit pleinement efficace ici.

### Enseignements clés

1. **Un modèle simple bien conçu surpasse souvent des architectures complexes** sur des tâches bien délimitées.
2. **Le transfer learning n'est pas toujours meilleur** : la distance entre le domaine source et le domaine cible est un facteur déterminant.
3. **Le LSTM n'apporte pas de valeur** sur des textes courts où les dépendances séquentielles longues sont rares.
4. **Le F1-score est la bonne métrique** sur un dataset déséquilibré : l'accuracy seule est trompeuse (un modèle naïf "tout ham" aurait déjà 86.6% d'accuracy).

### Réponse à la question métier

> **Peut-on détecter efficacement un SMS spam à partir du seul texte du message ?**

**Oui.** Le Modèle 1 atteint un **F1-score de 0.9448** et un **recall de 0.9195** sur la classe spam, ce qui signifie que plus de **9 spams sur 10 sont correctement détectés**. Avec seulement 12 faux positifs sur 1 115 messages, l'expérience utilisateur reste excellente.

Ce modèle constitue une base solide et directement déployable pour un système de filtrage automatique chez AT&T.
